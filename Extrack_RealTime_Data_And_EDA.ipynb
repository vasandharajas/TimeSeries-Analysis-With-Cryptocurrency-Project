# Exploratory Data Analysis

  1. Bitcoin 2014 - 2025 Dataset --- EDA

```python
#### To Visualizing the Plot
cleaned_btc_data = pd.read_csv(r"your Cleaned_bitcoin_2014_2025.csv")
cleaned_btc_data['Date'] = pd.to_datetime(cleaned_btc_data['Date'])
cleaned_btc_data.set_index('Date', inplace=True)

# Trend Visualizing
plt.figure(figsize=(12,5))
plt.plot(cleaned_btc_data.index, cleaned_btc_data['Close'])
plt.title("Bitcoin Price Trend (2014–2025)")
plt.xlabel("Year")
plt.ylabel("Close Price")
plt.show()
```

```python
#### Volume Trend Visualizing
plt.figure(figsize=(10,4))
plt.plot(cleaned_btc_data.index, cleaned_btc_data['Volume'])
plt.title("Trading Volume Over Time")
plt.show()
```

  # 2. Btc_Usd Visualizing

```python
## To Visualizing the BTC_USD
cleaned_btc_usd = pd.read_csv(r"your Cleaned_bit_usd_data.csv", header=[0,1])
cleaned_btc_usd.columns = cleaned_btc_usd.columns.get_level_values(0)
cleaned_btc_usd['Date'] = pd.to_datetime(cleaned_btc_usd['Date'])
cleaned_btc_usd.set_index('Date', inplace=True)

## Price Trend Visualizing
plt.figure(figsize=(12,5))
sns.lineplot(data=cleaned_btc_usd, x=cleaned_btc_usd.index, y='Close')
plt.title("BTC_USD Price Trend")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
##### Volatility Visualizing
btc_usd['returns'] = btc_usd['Close'].pct_change()
sns.histplot(btc_usd['returns'].dropna(), bins=50, kde=True)
plt.title("BTC Returns Distribution")
plt.show()
```

  # 3. Cryptocurrency EDA --- Visualizing

```python
crypto = pd.read_csv(r"your Cleaned_Cryptocurrency.csv")

### Top Coins Trend
top = crypto[crypto['name'].isin(['Bitcoin', 'Ethereum', 'Chainlink'])]
plt.figure(figsize=(12,5))
sns.lineplot(data=top, x='timestamp', y='price_usd', hue='name')
plt.title("Top Crypto Price Trends")
plt.show()
```

```python
## Clean Numeric Column
cols_money = ['price_usd','vol_24h','total_vol','market_cap']
for col in cols_money:
    crypto[col] = (
        crypto[col]
        .astype(str)  
        .str.replace(r'[\$,]', '', regex=True)  
        .str.replace('%', '', regex=False)  
    )
    crypto[col] = pd.to_numeric(crypto[col], errors='coerce')  

cols_percent = ['chg_24h','chg_7d']
for col in cols_percent:
    crypto[col] = (
        crypto[col]
        .astype(str)
        .str.replace('%', '', regex=False)
    )
    crypto[col] = pd.to_numeric(crypto[col], errors='coerce')
```

```python
### Correlation HeatMap
plt.figure(figsize=(10,6))
sns.heatmap(crypto.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Crypto Feature Correlation")
plt.show()
```
